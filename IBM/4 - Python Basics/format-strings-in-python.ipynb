# String interpolation (f-string)
Introduced in Python 3.6, f-strings are a new way to format strings in Python. They are prefixed with 'f' and use curly braces to enclose variables:

```python
name = "John"
age = 30
print(f"My name is {name} and I am {age} years old.")
```

# str.format()
This is another way to format strings in Python:

```python
name = "John"
age = 50
print("my name is {} and I am {} years old.".format(name, age))
```

# % operator
This is one of the oldest ways to foramt strings in Python:

```python
name = "Johnathan"
age = 30
print("My name is %s and I am %d years old." % (name, age))
```

**%(name,age):** This is a tuple containing the variables name and age. These variables replace the placeholders in the string.

# Additional capabilities
F-strings are also capable of evaluating expressions in curly braces:

```python
x = 10
y = 20
print(f"The sum of x and y is {x+y}.")
```

# Raw String (r")
Raw strings are powerful tools for handling textual data, especially when dealing with escape characters. Python treats the string as raw, meaning it interprets backslashes as literal characters.

```python
regular_string = "C:\new_folder\file.txt"
print("Regular String", regular_string)
```

```python
raw_string = r"C:\new_folder\file.txt";
print(raw_string)
```
