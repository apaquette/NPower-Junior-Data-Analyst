## Identify and Analyze Duplicates


### Task 1: Identify Duplicate Rows
1. Count the number of duplicate rows in the dataset.
3. Display the first few duplicate rows to understand their structure.


```python
## Write your code here
dup_count = df.duplicated().sum()
print("Duplicated count:",dup_count)

duplicates = df[df.duplicated()]
duplicates.head()
```

### Task 2: Analyze Characteristics of Duplicates
1. Identify duplicate rows based on selected columns such as MainBranch, Employment, and RemoteWork. Analyse which columns frequently contain identical values within these duplicate rows.
2. Analyse the characteristics of rows that are duplicates based on a subset of columns, such as MainBranch, Employment, and RemoteWork. Determine which columns frequently have identical values across these rows.
   


```python
## Write your code here
duplicated_subset = duplicates[['MainBranch', 'Employment', 'RemoteWork']]
print('Identical values across rows:',duplicated_subset.duplicated().sum())
duplicated_subset
```

### Task 3: Visualize Duplicates Distribution
1. Create visualizations to show the distribution of duplicates across different categories.
2. Use bar charts or pie charts to represent the distribution of duplicates by Country and Employment.


```python
## Write your code here
dupe_country_counts = duplicates['Country'].value_counts()

# make some countries more readable
dupe_country_counts.index = dupe_country_counts.index.str.replace('United States of America', 'USA')
dupe_country_counts.index = dupe_country_counts.index.str.replace('United Kingdom of Great Britain and Northern Ireland', 'UK')
dupe_country_counts.index = dupe_country_counts.index.str.replace('Iran, Islamic Republic of...', 'Iran')

dupe_employment_counts = duplicates['Employment'].value_counts()

fig = plt.figure(figsize=(12,6))
ax0 = fig.add_subplot(1,2,1)
ax1 = fig.add_subplot(1,2,2)

dupe_country_counts.plot(kind='bar', ax=ax0)
ax0.set_title('Duplicates by Country')
ax0.set_xlabel('Country')
ax0.set_ylabel('Duplicate Count')

dupe_employment_counts.plot(kind='bar', ax=ax1)
ax1.set_title('Duplicates by Employment Type')
ax1.set_xlabel('Employment')


plt.show()
```

### Task 4: Strategic Removal of Duplicates
1. Decide which columns are critical for defining uniqueness in the dataset.
2. Remove duplicates based on a subset of columns if complete row duplication is not a good criterion.


```python
## Write your code here
df_deduped = df.drop_duplicates(
    subset=["ResponseId"],
    keep="first"
)
df_deduped['ResponseId'].is_unique
```

## Verify and Document Duplicate Removal Process


### Task 5: Documentation
1. Document the process of identifying and removing duplicates.


# Write your explanation here
Given that the dataset is a collection of survey responses, the only criteria that can reliably identify duplicates is the ResponseId column. This column uniquely identifies each response, and duplicated Ids mean there are duplicate responses. This means our main uniqueness identifyer will be `df['ResponseId'].is_unique`. If this returns false, it means we have some duplicated responses. Removing duplicates is as simple as running the following:

`df_deduped = df.drop_duplicates(subset['ResponseId'], keep='First')`

We then executed `df['ResponseId'].is_unique` again to validate the duplicates were removed successfully.
