4. Top 10 Web Framework Worked With

Import the required libraries

```python
import pandas as pd
```

Import the dataset

```python
# load dataset
df = pd.read_csv('survey_data_updated.csv')
```

Create a function that takes the dataframe and a target column, and returns a dataframe with the top 10 entries from that target column. This method is specific to the `survey_data_updated.csv` file, which needs to explore the dataframe based on the list of technologies in the specific columns.

```python
def process_top_10(df: pd.DataFrame, target:str) -> pd.DataFrame:
    df[target] = df[target].str.split(';').explode(target)
    top10 = df[target].value_counts().head(10).index
    df_top = df[df[target].isin(top10)]
    df_top10 = pd.DataFrame(df_top.groupby(target)['ResponseId'].count().reset_index())
    df_top10.rename(columns={'ResponseId': 'Count'}, inplace=True)
    return df_top10
```

Create arrays containing the top 10 columns needing targeting, the renaming of the column, and expected filename output when processing is completed.

```python
# process data to obtain top 10 languages worked with

top10_cols = ['LanguageHaveWorkedWith', 'DatabaseHaveWorkedWith', 'PlatformHaveWorkedWith', 'WebframeHaveWorkedWith']
col_rename = ['Languages', 'Databases', 'Platforms', 'Web Frameworks']
file_name = ['top_10_programming_languages', 'top_10_databases', 'top_10_platforms', 'top_10_webframeworks']

for index in range(0,4):
    target_col = top10_cols[index]
    df_top10 = process_top_10(df, target_col)
    df_top10.rename(columns={target_col: col_rename[index]}, inplace=True)
    df_top10.to_csv(f'{file_name[index]}.csv', index=False)
```

For Education, we need to rename many of the categories to be shorter form so it's easier to read in visualizations. Same as we did for countries.

```python
df['EdLevel'] = df['EdLevel'].replace("Bachelor's degree (B.A., B.S., B.Eng., etc.)", "Bachelor's degree")
df['EdLevel'] = df['EdLevel'].replace("Master's degree (M.A., M.S., M.Eng., MBA, etc.)", "Master's degree")
df['EdLevel'] = df['EdLevel'].replace("Professional degree (JD, MD, Ph.D, Ed.D, etc.)", "Professional degree")
df['EdLevel'] = df['EdLevel'].replace("Some college/university study without earning a degree", "Some college/university")
df['EdLevel'] = df['EdLevel'].replace("Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)", "Secondary school")
df['EdLevel'] = df['EdLevel'].replace("Associate degree (A.A., A.S., etc.)", "Associate degree")
df['EdLevel'] = df['EdLevel'].replace("Primary/elementary school", "Elementary school")
```

```python
respondent_by_education = df['EdLevel'].value_counts().reset_index()
respondent_by_education.to_csv('demographic_education.csv', index=False)
```
