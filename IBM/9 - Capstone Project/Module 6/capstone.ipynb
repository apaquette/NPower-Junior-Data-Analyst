```python
import seaborn as sns
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# import the job-postings file into a dataframe
df = pd.read_excel('job-postings.xlsx')
df.head()
```

```python
# Create a bar chart of the data
sns.barplot(data=df, x='Location', y='Job Listings', order=df.sort_values('Job Listings', ascending=False).Location)
plt.xticks(rotation=45)
plt.title('Job listings per Location')
plt.show()
```

```python
df = pd.read_csv('popular-languages.csv')
df.head()
```

```python
df['Average Annual Salary'] = df['Average Annual Salary'].replace({'\$': '', ',':''},regex=True).astype(int)
df.dtypes
```

```python
sns.barplot(data=df, x='Language', y='Average Annual Salary', order=df.sort_values('Average Annual Salary', ascending=False).Language)
plt.title('Programming Language per Salary')
plt.xticks(rotation=45)
plt.show()
```
